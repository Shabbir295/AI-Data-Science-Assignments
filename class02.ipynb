```python
import turtle
import colorsys

# Screen setup
screen = turtle.Screen()
screen.bgcolor("black")
screen.title("Rainbow Mandala 🌈")

# Turtle setup
t = turtle.Turtle()
t.speed(0)
t.width(2)
t.hideturtle()

# Rainbow colors + spiral pattern
hue = 0

for i in range(360):
    # HSV → RGB
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)

    # Turtle RGB mode
    t.pencolor(r, g, b)

    # Draw pattern
    t.forward(i * 1.2)
    t.right(59)

    # Change color
    hue += 0.008

# Keep window open
turtle.done()
```

```python
import turtle
import colorsys

# Screen
screen = turtle.Screen()
screen.setup(900, 600)
screen.bgcolor("black")
screen.title("Audi Rings - Turtle")

# Turtle
t = turtle.Turtle()
t.speed(0)
t.hideturtle()
t.width(8)

# RGB colors
screen.colormode(1.0)

# 4 interlocking rings
x_positions = [-180, -60, 60, 180]

for i, x in enumerate(x_positions):
    # Rainbow color using colorsys
    r, g, b = colorsys.hsv_to_rgb(i / 4, 1, 1)
    t.pencolor(r, g, b)

    t.penup()
    t.goto(x, 0)
    t.setheading(0)
    t.pendown()

    t.circle(90)

# Text
t.penup()
t.goto(0, -150)
t.pencolor("white")
t.write(
    "AUDI",
    align="center",
    font=("Arial", 28, "bold")
)

turtle.done()
```

```python
import turtle
import colorsys
import math
import tkinter as tk

# =========================
# SCREEN
# =========================

screen = turtle.Screen()
screen.setup(1000, 800)
screen.bgcolor("black")
screen.title("NEON GALAXY 🌌")

# Window ko front par lao
root = screen.getcanvas().winfo_toplevel()
root.deiconify()
root.lift()
root.attributes("-topmost", True)
root.after(500, lambda: root.attributes("-topmost", False))
root.focus_force()

screen.colormode(1.0)

# =========================
# TURTLE
# =========================

t = turtle.Turtle()
t.speed(0)
t.hideturtle()
t.width(2)

# =========================
# NEON GALAXY
# =========================

for i in range(720):

    # Rainbow color
    hue = (i % 360) / 360
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)

    t.pencolor(r, g, b)

    # Start from center
    t.penup()
    t.goto(0, 0)

    # Spiral angle
    angle = i * 7

    t.setheading(angle)

    t.pendown()

    # Spiral length
    length = 100 + i * 0.7

    # Draw curved galaxy arm
    for j in range(18):
        t.forward(length / 18)
        t.right(8)

# =========================
# CENTER ORBIT RINGS
# =========================

for ring in range(12):

    hue = (ring / 12)
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)

    t.pencolor(r, g, b)
    t.width(3)

    t.penup()
    t.goto(0, -ring * 8)
    t.setheading(0)
    t.pendown()

    t.circle(ring * 8)

# =========================
# TITLE
# =========================

t.penup()
t.goto(0, -350)

t.pencolor("white")
t.write(
    "NEON GALAXY",
    align="center",
    font=("Arial", 28, "bold")
)

turtle.done()
```

```python
import turtle
import colorsys
import math

# =========================
# SCREEN
# =========================

screen = turtle.Screen()
screen.setup(1000, 800)
screen.bgcolor("black")
screen.title("⚡ NEON HYPER PORTAL ⚡")
screen.colormode(1.0)

# Window ko saamne lao
root = screen.getcanvas().winfo_toplevel()
root.deiconify()
root.lift()
root.attributes("-topmost", True)
root.after(700, lambda: root.attributes("-topmost", False))
root.focus_force()

# =========================
# TURTLE
# =========================

t = turtle.Turtle()
t.speed(0)
t.hideturtle()
t.width(2)

# =========================
# HYPER PORTAL
# =========================

layers = 180

for i in range(layers):

    # Rainbow color
    hue = (i / layers) % 1.0
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)

    t.pencolor(r, g, b)

    # Radius
    radius = 20 + i * 2.2

    # Start position
    t.penup()
    t.goto(0, 0)

    # Rotation
    t.setheading(i * 13)

    t.pendown()

    # Polygon / portal shape
    sides = 7

    for j in range(sides):

        # Slight mathematical distortion
        angle = (i * 3 + j * 360 / sides)

        x = radius * math.cos(math.radians(angle))
        y = radius * math.sin(math.radians(angle))

        t.goto(x, y)

# =========================
# SECOND VORTEX
# =========================

t.width(1)

for i in range(250):

    hue = (0.75 + i / 1000) % 1.0
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)

    t.pencolor(r, g, b)

    t.penup()
    t.goto(0, 0)

    t.setheading(i * 9)

    t.pendown()

    length = 50 + i * 1.4

    for j in range(8):

        t.forward(length / 8)
        t.right(45)

# =========================
# CENTER CORE
# =========================

for i in range(40):

    hue = i / 40
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)

    t.pencolor(r, g, b)
    t.width(3)

    t.penup()
    t.goto(0, -i * 3)
    t.setheading(0)
    t.pendown()

    t.circle(i * 3)

# =========================
# TITLE
# =========================

t.penup()
t.goto(0, -350)

t.pencolor("white")
t.write(
    "HYPER PORTAL",
    align="center",
    font=("Arial", 30, "bold")
)

turtle.done()
```
